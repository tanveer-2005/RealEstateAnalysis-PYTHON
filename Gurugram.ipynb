```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data of gurugram real Estate.csv")
print(df)
```

```python
print(df.columns.tolist())
```

```python
print(df.info())
```

```python
# DATA CLEANING
# 
df.columns = df.columns.str.strip().str.lower().str.replace(' ','_')
```

```python
print(df.columns.tolist())
```

```python
df.drop_duplicates()
```

```python
# NUUMERIC COLUMNS CLEANING
df["price"]=df["price"].astype(str).str.replace(',','').astype(float)
```

```python
df["area"]=df["area"].astype(str).str.replace(',','').astype(int)
```

```python
df["rate_per_sqft"]=df["rate_per_sqft"].astype(str).str.replace(',','').astype(float).astype(int)
```

```python
# CATEGORICAL CLEANING
```

```python
df["status"] = df["status"].str.lower().str.strip()
```

```python
df["rera_approval"] = df["rera_approval"].str.lower().str.strip().map({'approved by rera': True, 'not approved by rera':False})
```

```python
print(df["rera_approval"])
```

```python
df["flat_type"] = df["flat_type"].str.lower().str.strip()
```

```python
df = df.drop_duplicates()
print(df)
```

```python
print(df.columns.tolist)
```

```python
## Which is the costliest flat in the dataset???
print(df.loc[df["price"].idxmax()])
```

```python
# WHich locality has the highest average price?
print(df.groupby("locality")["price"].mean().idxmax())
```

```python
# Which Locality has the highest rate per square feet?
print(df.groupby("locality")["rate_per_sqft"].mean().idxmax())
```

```python
# Do ready tomove properties cost more than under construction properties?
print("Ready To Move : ", df[df["status"]=="ready to move"]["price"].mean())
print("Under Construction : ",df[df["status"]=="under construction"]["price"].mean())
```

```python
# Do rera approved property commann price premium
print("Ready To Move : ", df[df["rera_approval"]==True]["price"].mean())
print("Under Construction : ",df[df["rera_approval"]==False]["price"].mean())
```

```python
# How does area sqft impact price property ?
sns.scatterplot(x = "area", y = "price", data = df)
plt.title("AREA IMPACT")
plt.xlabel("AREA")
plt.ylabel("PRICE")
plt.show()
```

```python
# Which BHK configuration is the most expensive?
print(df.groupby("bhk_count")["rate_per_sqft"].mean().idxmax())
```

```python
# Which BHK configuration is the most expensive?
print(df.groupby("flat_type")["rate_per_sqft"].mean().idxmax())
```

```python
# DO certain builder price higher
print(df.groupby("company_name")["rate_per_sqft"].mean().sort_values(ascending=False).head(5))
```

```python
# Are larger homes more expensive per sqft?
sns.scatterplot(x = "area", y = "rate_per_sqft", data = df)
plt.title("AREA IMPACT")
plt.xlabel("AREA")
plt.ylabel("SQFT")
plt.show()
```
